```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import Select
import time
import os

#Define download directory
download_dir = os.path.abspath("../data/raw")

#Set up headless Chrome options
options = Options()
options.add_argument('--headless')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
options.add_experimental_option("prefs", {
    "download.default_directory": download_dir,
    "download.prompt_for_download": False,
    "directory_upgrade": True,
    "safebrowsing.enabled": True
})

#Launch browser
driver = webdriver.Chrome(options=options)
driver.get("https://www.transtats.bts.gov/OT_Delay/OT_DelayCause1.asp?pn=1")

#STEP 1: Select "January, 2020" from PeriodFrom
#Can modify select by value for earlier start dates
Select(driver.find_element("id", "PeriodFrom")).select_by_value("24241")  

#STEP 2: Click the submit button using its ID
driver.find_element("id", "submit").click()
time.sleep(2)

#STEP 3: Click the "Download Raw Data" link
download_link = driver.find_element("link text", "Download Raw Data")
download_link.click()

#Wait for download
time.sleep(15)
driver.quit()

print("? Download complete: Delay summary data January 2020 to Present saved to data/raw/")
```
